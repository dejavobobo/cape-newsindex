# EPU index from a real article archive

Runs `CAPEnewsindex.build_index` on a scraped archive laid out as
`<archive>/<country>/<newspaper>/<YYYY-MM>/articles_*.csv.gz` -- e.g. a Google Drive folder
synced to this computer, or mounted in Colab. Edit the settings, run all cells.

The first run on a cloud folder is slower while files download; later runs reuse the cache.

```python
# In Colab:  %pip install -q CAPEnewsindex[plot,excel]
#            from google.colab import drive; drive.mount("/content/drive")
import CAPEnewsindex as ni

ARCHIVE   = r"C:\path\to\Scrapped_Top_Articles"     # or "/content/drive/MyDrive/..." in Colab
KEYWORDS  = "my_keywords.xlsx"                          # or .csv / .txt / .json
COUNTRIES = ["Nigeria", "Ghana"]
START, END = "2024-01", None                            # None = latest month in the archive
HISTORY   = None                                          # e.g. ("2015-01", "2023-12")
GDP       = None                                          # your GDP file -> group EPU
GROUPS    = None                                          # e.g. {"West Africa": ["Nigeria", "Ghana"]}
```

```python
# No keyword file yet? Write the production keywords for these countries as one to edit:
# ni.keywords_template(KEYWORDS, countries=COUNTRIES)
```

```python
result = ni.build_index(ARCHIVE, KEYWORDS, COUNTRIES, START, END,
                        history=HISTORY, gdp=GDP, groups=GROUPS, output="my_results")
result.show()
```

```python
result.plot();
```

`my_results/epu_dashboard.html` (open it in a browser) and `my_results/epu_results.xlsx` now hold every chart and table.
